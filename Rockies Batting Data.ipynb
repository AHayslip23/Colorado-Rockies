**Classification How Accurately Can I predict whether a rockies player is left handed (68.5% Accuracy)**

```python
classification = rockies_batting[["Plate_Appearances", "At_Bats", "Runs", "Hits", "Doubles", "Triples", "Home_Runs", "Runs_Batted_In", "Base_On_Balls", "Batting_Average", "On_Base_Percentage", "Slugging_Percentage", "On_Base_Plus_Slugging_Percentage", "On_Base_Plus_Slugging_Percentage_Plus", "Total_Bases",  "Left Handed Batter"]]
```

```python
X = classification.iloc[:, :-1].values
Y = classification.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test= train_test_split(X, Y, random_state=1, test_size=0.25, shuffle=True)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train, Y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
cm = confusion_matrix(Y_test, y_pred)
print(cm)
accuracy_score(Y_test, y_pred)
```

**Try to do Logistic Regression Cross Validation (71.5% accuracy)**

```python
from sklearn.model_selection import train_test_split 
import pickle
from sklearn.linear_model import LogisticRegressionCV

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=1, test_size=0.25, shuffle=True)

clf = LogisticRegressionCV(cv=10,
                          scoring="accuracy",
                          random_state=0,
                          n_jobs=-1,
                          verbose=3,
                          max_iter=400).fit(X_train, Y_train)
```

```python
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, y_pred)
print(cm)
accuracy_score(Y_test, y_pred)
```
