**Create functions to generate new columns for quality of a pitchers performance based on their era, whip and fip strikeout per nine innings and walks per nine innings**

```python
def era_quality(row):
    era = row["Earned_Run_Average"]
    if era < 2.00:
        return "Outstanding"
    elif 2.00 <= era < 3.00:
        return "Awesome"
    elif 3.00 <= era < 4.00:
        return "Good"
    elif 4.00 <= era < 5.00:
        return "Average"
    else:
        return "Below Average" 
```

```python
rockies_pitching["ERA Performance"] = rockies_pitching.apply(era_quality, axis=1)
```

```python
def whip_quality(row):
    whip = row["Walks_Hits_Per_Inning_Pitched"]
    if 0.80 <= whip < 1.00:
        return "Excellent"
    elif 1.00 <= whip < 1.10:
        return "Great"
    elif 1.10 <= whip < 1.25:
        return "Above Average"
    elif 1.25 <= whip < 1.32:
        return "Average"
    elif 1.32 <= whip < 1.40:
        return "Below Average"
    elif 1.40 <= whip < 1.50:
        return "Poor"
    else:
        return "Awful"
```

```python
rockies_pitching["WHIP Performance"] = rockies_pitching.apply(whip_quality, axis=1)
```

```python
def fip_quality(row):
    fip = row["Fielding_Independent_Pitching"]
    if fip < 2.00:
        return "Outstanding"
    elif 2.00 <= fip < 3.00:
        return "Awesome"
    elif 3.00 <= fip < 4.00:
        return "Good"
    elif 4.00 <= fip < 5.00:
        return "Average"
    else:
        return "Below Average" 
```

```python
rockies_pitching["FIP Performance"] = rockies_pitching.apply(fip_quality, axis=1)
```

**Model 1 Hits Allowed and Home Runs Allowed**

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=0, shuffle=True)
```

**Cross Validation for Linear Regression**

**Perform a 5 fold cross validation increases the score slighly from 73.5% for the base model to 75.2%**

```python
from sklearn.model_selection import KFold, cross_val_score
```

```python
n_splits = 5
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
regressor = LinearRegression()
```

```python
scores = cross_val_score(regressor, X_train, Y_train, cv=kf, scoring='r2')
best_r2_score = np.max(scores)
```

```python
print(f"R-squared scores for each fold: {scores}")
print(f"Best R-squared score: {best_r2_score}")
```
